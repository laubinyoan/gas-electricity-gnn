```python
#!/usr/bin/env python3
"""
GNN Couplé Gaz/Électricité — Réseau Énergétique Belge 2024
Implémentation complète en un seul script exécutable.
"""
 
import os, json, random
from dataclasses import dataclass
 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
 
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler
 
# ════════════════════════════════════════════════════════════════════════════
# CONFIG
# ════════════════════════════════════════════════════════════════════════════
 
DATA_DIR = "/mnt/user-data/uploads"
 
@dataclass
class Config:
    gas_network_path    : str = f"./Gas Network Data.xlsx"
    gas_cons_ind_path   : str = f"./input/Gas Cons Ind.xlsx"
    gas_cons_distrib_path:str = f"./input/Gas Cons Distrib.xlsx"
    gas_cons_export_path: str = f"./input/Gas Cons Export.xlsx"
    decision_vars_path  : str = f"./output/Decision_variablesRU.xlsx"
    elec_network_path   : str = f"./Elec Network Data.xlsx"
    elec_load_path      : str = f"./input/Elec Load Data.xlsx"
    wind_path           : str = f"./input/Wind Data.xlsx"
    output_dir          : str = "/mnt/user-data/outputs/coupled_gnn"
 
    num_workers : int   = 0   # ← ajouter cette ligne
    num_gas_nodes       : int   = 28
    num_elec_nodes      : int   = 45
    num_generators      : int   = 24
    gas_well_nodes_1based: tuple = (1, 5, 17, 18)
 
    input_len   : int   = 72
    output_len  : int   = 24
    train_ratio : float = 0.70
    val_ratio   : float = 0.15
    test_ratio  : float = 0.15
 
    hidden_dim  : int   = 128
    gnn_hidden  : int   = 64
    edge_hidden : int   = 32
    emb_dim     : int   = 16
    dropout     : float = 0.2
 
    batch_size   : int   = 64
    lr           : float = 3e-4
    weight_decay : float = 5e-4
    max_epochs   : int   = 200
    patience     : int   = 30
    grad_clip    : float = 1.0
    alpha        : float = 0.65
 
    seed : int = 42
 
CFG = Config()
os.makedirs(CFG.output_dir, exist_ok=True)
 
def set_seed(seed=42):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
 
set_seed(CFG.seed)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
 
```

```python
# ════════════════════════════════════════════════════════════════════════════
# UTILS
# ════════════════════════════════════════════════════════════════════════════
 
def impute_numeric_df(df):
    out = df.copy()
    for c in out.columns:
        out[c] = pd.to_numeric(out[c], errors="coerce").fillna(0)
    return out
 
def read_monthly_decision(path, prefix):
    xls = pd.ExcelFile(path)
    sheets = sorted([s for s in xls.sheet_names if s.startswith(prefix)],
                    key=lambda s: int(s.split("-")[1]))
    dfs = []
    for s in sheets:
        df = pd.read_excel(path, sheet_name=s)
        dfs.append(df.iloc[:, 1:].reset_index(drop=True))
    return pd.concat(dfs, axis=0, ignore_index=True)
 
def read_monthly_gas_exo(path):
    xls = pd.ExcelFile(path)
    sheets = sorted([s for s in xls.sheet_names if s != "year"],
                    key=lambda s: int(s.split("-")[0]))
    dfs = []
    for s in sheets:
        dfs.append(pd.read_excel(path, sheet_name=s).reset_index(drop=True))
    return pd.concat(dfs, axis=0, ignore_index=True)
```

```python
# ════════════════════════════════════════════════════════════════════════════
# CHARGEMENT DES DONNÉES
# ════════════════════════════════════════════════════════════════════════════
 
print("\n=== Chargement données gaz ===")
g_raw      = read_monthly_decision(CFG.decision_vars_path, "g-").astype(np.float32)
T          = len(g_raw)
dt_series  = pd.date_range(start="2024-01-01", periods=T, freq="h")
print(f"g_raw : {g_raw.shape}  T={T}  {dt_series[0]} → {dt_series[-1]}")
 
ind_df     = read_monthly_gas_exo(CFG.gas_cons_ind_path)
distrib_df = read_monthly_gas_exo(CFG.gas_cons_distrib_path)
export_df  = read_monthly_gas_exo(CFG.gas_cons_export_path)
for df_name, df_obj in [("ind", ind_df), ("distrib", distrib_df), ("export", export_df)]:
    df_obj["Date Time"] = pd.to_datetime(df_obj["Date Time"])
    df_obj.drop(df_obj[df_obj["Date Time"] < dt_series[0]].index, inplace=True)
    df_obj.drop(df_obj[df_obj["Date Time"] > dt_series[-1]].index, inplace=True)
    df_obj.reset_index(drop=True, inplace=True)
 
print(f"Exo lengths: ind={len(ind_df)} distrib={len(distrib_df)} export={len(export_df)}")
 
nodes_gas_df = pd.read_excel(CFG.gas_network_path, sheet_name="Nodes")
pipes_df     = pd.read_excel(CFG.gas_network_path, sheet_name="Pipelines")
well_df      = pd.read_excel(CFG.gas_network_path, sheet_name="Well")
pipes_df     = impute_numeric_df(pipes_df[[
    "Pipeline","From node","To node",
    "Natural gas flow constant [MNm3/(bar*h)2]","Compressor factor",
    "High Linepack","Initial linepack [MNm3]","Linepack constant [MNm3/bar]",
    "Estimated lenght [km]"
]])
for c in ["Res_Load_share_%","Ind_Load_share_%","Exp_Load_share_%"]:
    nodes_gas_df[c] = pd.to_numeric(
        nodes_gas_df[c].astype(str).str.replace(",","."), errors="coerce"
    ).fillna(0)
 
print("\n=== Chargement données électricité ===")
elec_load_raw = pd.read_excel(CFG.elec_load_path, sheet_name="Year 2024")
elec_load_raw["Datetime"] = pd.to_datetime(
    elec_load_raw["Datetime"].astype(str).str.replace(r"\+0[12]:00$","",regex=True))
elec_load_arr = elec_load_raw["Measured & Upscaled [MW]"].to_numpy(dtype=np.float32)[:T]
 
wind_raw = pd.read_excel(CFG.wind_path, sheet_name="Year 2024")
wind_raw["Datetime"] = pd.to_datetime(
    wind_raw["Datetime"].astype(str).str.replace(r"\+0[12]:00$","",regex=True))
wind_arr = wind_raw["Measured & Upscaled [MW]"].to_numpy(dtype=np.float32)[:T]
 
p_raw      = read_monthly_decision(CFG.decision_vars_path, "p-").astype(np.float32)
gen_df     = pd.read_excel(CFG.elec_network_path, sheet_name="Gen")
pmax_arr   = gen_df["Pmax_MW"].to_numpy(dtype=np.float32)
# p_raw is in units of 100 MW (p_raw * 100 = MW). Use directly.
# p_raw_denorm = p_raw.to_numpy().astype(np.float32)  <- NO Pmax multiplication
Y_elec_raw_unit = p_raw.to_numpy().astype(np.float32)  # in 100-MW units
p_raw_denorm = Y_elec_raw_unit  # alias for compatibility
 
buses_df   = pd.read_excel(CFG.elec_network_path, sheet_name="Buses")
lines_df   = pd.read_excel(CFG.elec_network_path, sheet_name="LinesMerged")
genmap_df  = pd.read_excel(CFG.elec_network_path, sheet_name="GenMap")
windmap_df = pd.read_excel(CFG.elec_network_path, sheet_name="WindMap")
 
gen_bus_idx   = (gen_df["Node"].astype(int) - 1).to_numpy()
gas_gen       = gen_df[gen_df["Gas node"].notna()].copy()
coup_gas_idx  = (gas_gen["Gas node"].astype(int) - 1).to_numpy()
coup_elec_idx = (gas_gen["Node"].astype(int)  - 1).to_numpy()
coup_eff      = gas_gen["Efficiency"].to_numpy(dtype=np.float32)
coup_pmax_norm = (gas_gen["Pmax_MW"].to_numpy(dtype=np.float32) / gas_gen["Pmax_MW"].max())
 
print(f"p_raw_denorm : {p_raw_denorm.shape}  range {p_raw_denorm.min():.1f}..{p_raw_denorm.max():.1f} MW")
print(f"Coupling pairs: {len(coup_gas_idx)}")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# CONSTRUCTION DES GRAPHES
# ════════════════════════════════════════════════════════════════════════════
 
print("\n=== Construction graphes ===")
 
# --- Graphe gaz ---
def safe_float(val, default=0.):
    if pd.isna(val): return default
    try:
        return float(str(val).strip().split("/")[0].strip())
    except ValueError:
        return default

def build_gas_graph():
    N = CFG.num_gas_nodes
    edge_cols = [
        "Natural gas flow constant [MNm3/(bar*h)2]", "Compressor factor",
        "High Linepack", "Initial linepack [MNm3]",
        "Linepack constant [MNm3/bar]", "Estimated lenght [km]",
    ]
    ea_df = impute_numeric_df(pipes_df[edge_cols])
    ep, ea = [], []
    for i, row in pipes_df.iterrows():
        u, v = int(row["From node"])-1, int(row["To node"])-1
        a = ea_df.iloc[i].to_numpy(dtype=np.float32)
        ep += [(u,v),(v,u)]; ea += [a,a]

    edge_index    = np.array(ep, dtype=np.int64).T
    edge_attr_raw = np.array(ea, dtype=np.float32)
    edge_scaler   = StandardScaler()
    edge_attr     = edge_scaler.fit_transform(edge_attr_raw).astype(np.float32)

    ndf   = nodes_gas_df.copy().sort_values("Gas Node").reset_index(drop=True)
    nids  = ndf["Gas Node"].astype(int).to_numpy()
    res_s = ndf["Res_Load_share_%"].to_numpy(dtype=np.float32)
    ind_s = ndf["Ind_Load_share_%"].to_numpy(dtype=np.float32)
    exp_s = ndf["Exp_Load_share_%"].to_numpy(dtype=np.float32)
    pr_mn = ndf["Pr_min [bar]"].to_numpy(dtype=np.float32)
    pr_mx = ndf["Pr_max [bar]"].to_numpy(dtype=np.float32)

    deg = np.zeros(N, dtype=np.float32)
    for u,v in pipes_df[["From node","To node"]].astype(int).to_numpy():
        deg[u-1]+=1; deg[v-1]+=1

    well_set = set(well_df["Gas node"].dropna().astype(int))
    comp_set = set(pipes_df[pipes_df["Compressor factor"]>1]
                   [["From node","To node"]].astype(int).values.flatten())
    exp_set  = {10,21,22,25}

    is_w  = np.array([1. if n in well_set else 0. for n in nids], dtype=np.float32)
    is_c  = np.array([1. if n in comp_set else 0. for n in nids], dtype=np.float32)
    is_ex = np.array([1. if n in exp_set  else 0. for n in nids], dtype=np.float32)
    is_s  = is_w.copy()
    is_ld = ((res_s>0)|(ind_s>0)|(exp_s>0)|(is_ex>0)).astype(np.float32)
    is_int= ((is_s==0)&(is_ld==0)).astype(np.float32)
    nid_n = ((nids-1)/(N-1)).astype(np.float32)

    gmax = np.zeros(N, dtype=np.float32)
    for _, r in well_df.iterrows():
        if pd.notna(r["Gas node"]) and pd.notna(r["Gmax [MNm3/h]"]):
            gmax[int(r["Gas node"])-1] = float(r["Gmax [MNm3/h]"])
    gmax_n = (gmax/gmax.max()).astype(np.float32)

    static = np.column_stack([
        nid_n, res_s, ind_s, exp_s, pr_mn, pr_mx, deg,
        is_w, np.zeros(N, dtype=np.float32),
        is_ex, is_c, is_s, is_ld, is_int, gmax_n
    ]).astype(np.float32)

    print(f"Gas  graph : edge_index={edge_index.shape}  "
          f"edge_attr={edge_attr.shape}  static={static.shape}")
    return dict(edge_index=edge_index, edge_attr=edge_attr,
                edge_attr_raw=edge_attr_raw, edge_scaler=edge_scaler,
                static=static, edge_dim=6, static_dim=15)


graph_gas = build_gas_graph()
print("graph_gas OK :", list(graph_gas.keys()))
 
# --- Graphe élec ---
def safe_float(val, default=0.):
    if pd.isna(val): return default
    try:
        return float(str(val).strip().split("/")[0].strip())
    except ValueError:
        return default

def build_elec_graph():
    N_E = CFG.num_elec_nodes
    ep, ea = [], []
    for _, row in lines_df.iterrows():
        u, v = int(row["FromNode"])-1, int(row["ToNode"])-1
        a = np.array([
            safe_float(row["X_pu"],      0.),
            safe_float(row["Pmax_MW"],   0.),
            safe_float(row["UB_kV"],   380.),
            safe_float(row["Length_km"], 50.),
        ], dtype=np.float32)
        ep += [(u,v),(v,u)]; ea += [a,a]

    edge_index_elec = np.array(ep, dtype=np.int64).T
    edge_attr_raw   = np.array(ea, dtype=np.float32)
    edge_scaler     = StandardScaler()
    edge_attr_elec  = edge_scaler.fit_transform(edge_attr_raw).astype(np.float32)

    buses = buses_df.copy().sort_values("Node").reset_index(drop=True)
    ls  = buses["Load_share_%"].to_numpy(dtype=np.float32)/100.
    pv  = buses["Pv_share_%"].to_numpy(dtype=np.float32)/100.
    on  = buses["Onshore_share_%"].to_numpy(dtype=np.float32)/100.
    ub  = (buses["UB_kV"].to_numpy(dtype=np.float32)/380.).astype(np.float32)
    nid = np.arange(N_E,dtype=np.float32)/(N_E-1)
    deg = np.zeros(N_E,dtype=np.float32)
    for u,v in lines_df[["FromNode","ToNode"]].astype(int).to_numpy():
        deg[u-1]+=1; deg[v-1]+=1
    deg_n    = (deg/deg.max()).astype(np.float32)
    gmap_a   = genmap_df.to_numpy(dtype=np.float32)
    wmap_a   = windmap_df["Offshore"].to_numpy(dtype=np.float32)
    is_ld_e  = (ls>0).astype(np.float32)
    is_gen_e = (gmap_a.sum(axis=1)>0).astype(np.float32)
    is_win_e = (wmap_a>0).astype(np.float32)
    is_cou_e = np.zeros(N_E,dtype=np.float32)
    for idx in coup_elec_idx: is_cou_e[idx]=1.

    static = np.column_stack([nid,ls,pv,on,ub,is_ld_e,
                               is_gen_e,is_win_e,is_cou_e,deg_n]).astype(np.float32)
    print(f"Elec graph : edge_index={edge_index_elec.shape}  edge_attr={edge_attr_elec.shape}  static={static.shape}")
    return dict(edge_index=edge_index_elec, edge_attr=edge_attr_elec,
                edge_attr_raw=edge_attr_raw, edge_scaler=edge_scaler,
                static=static, genmap=gmap_a, edge_dim=4, static_dim=10)

graph_elec = build_elec_graph()
```

```python
# ════════════════════════════════════════════════════════════════════════════
# CONSTRUCTION DES FEATURES NODALES
# ════════════════════════════════════════════════════════════════════════════
 
print("\n=== Construction features ===")
 
target_idx0 = [n-1 for n in CFG.gas_well_nodes_1based]
 
# --- Features gaz ---
# ─── CELL 10 — Features nœuds gaz ────────────────────────────────────────────

def build_gas_features():
    N   = CFG.num_gas_nodes
    tgt = [n-1 for n in CFG.gas_well_nodes_1based]   # [0,4,16,17]

    g_target = g_raw.to_numpy(dtype=np.float32)
    g_hist   = np.zeros((T, N), dtype=np.float32)
    for j, idx in enumerate(tgt): g_hist[:, idx] = g_target[:, j]

    g_lag1  = np.zeros_like(g_hist); g_lag1[1:]   = g_hist[:-1]
    g_lag2  = np.zeros_like(g_hist); g_lag2[2:]   = g_hist[:-2]
    g_lag3  = np.zeros_like(g_hist); g_lag3[3:]   = g_hist[:-3]
    g_lag24 = np.zeros_like(g_hist); g_lag24[24:] = g_hist[:-24]

    ind_g = ind_df["Physical Flow (MNm3)"].to_numpy(dtype=np.float32)
    dis_g = distrib_df["Physical Flow (MNm3)"].to_numpy(dtype=np.float32)

    # ── Shares nodaux depuis Gas Network Data ─────────────────────────────
    ndf = nodes_gas_df.copy().sort_values("Gas Node").reset_index(drop=True)
    res_share = ndf["Res_Load_share_%"].to_numpy(dtype=np.float32)  # [28]
    ind_share = ndf["Ind_Load_share_%"].to_numpy(dtype=np.float32)  # [28]

    # ── Signaux locaux : global × share ───────────────────────────────────
    # share=0 → local=0 (physiquement correct : nœud sans consommation)
    # Les puits (share=0) apprennent la demande via le message passing GNN
    # depuis leurs voisins consommateurs qui, eux, ont un share > 0
    ind_local  = (ind_g[:, None] * ind_share[None, :]).astype(np.float32)  # [T, 28]
    dis_local  = (dis_g[:, None] * res_share[None, :]).astype(np.float32)  # [T, 28]

    ir24 = pd.Series(ind_g).rolling(24, min_periods=1).mean().to_numpy(dtype=np.float32)
    dr24 = pd.Series(dis_g).rolling(24, min_periods=1).mean().to_numpy(dtype=np.float32)
    ir24_local = (ir24[:, None] * ind_share[None, :]).astype(np.float32)
    dr24_local = (dr24[:, None] * res_share[None, :]).astype(np.float32)

    # ── Vérification ──────────────────────────────────────────────────────
    print("Vérification ind_local au t=100 :")
    for node in [1, 2, 8, 17, 19]:
        idx = node - 1
        print(f"  Node {node:2d}  ind_share={ind_share[idx]:.4f}  "
              f"ind_local={ind_local[100, idx]:.4f}  "
              f"(global={ind_g[100]:.4f})")

    # ── Exports locaux ─────────────────────────────────────────────────────
    gcv  = export_df["GCV (kwh/Nm3)"].to_numpy(dtype=np.float32)
    bene = export_df["Physical Flow VIP BENE (kWh)"].to_numpy(dtype=np.float32)/gcv/1e6
    blar = export_df["Physical Flow Blaregnies L (kWh)"].to_numpy(dtype=np.float32)/gcv/1e6
    bras = export_df["Physical Flow Bras-Petange (kWh)"].to_numpy(dtype=np.float32)/gcv/1e6
    zpt  = export_df["Physical Flow VIP THE-ZPT (kWh)"].to_numpy(dtype=np.float32)/gcv/1e6

    exp_loc = np.zeros((T, N), dtype=np.float32)
    exp_loc[:, 9]  = bene
    exp_loc[:, 24] = blar
    exp_loc[:, 21] = bras
    exp_loc[:, 20] = zpt

    # ── Features temporelles ───────────────────────────────────────────────
    hour = pd.DatetimeIndex(dt_series).hour
    dow  = pd.DatetimeIndex(dt_series).dayofweek
    tf   = np.stack([hour/23., dow/6., (hour==23).astype(float)], -1)
    tf   = np.repeat(tf[:, None, :], N, 1).astype(np.float32)

    # ── Statiques répétées ─────────────────────────────────────────────────
    sf = np.repeat(graph_gas["static"][None, :, :], T, 0)

    # ── Stack final ────────────────────────────────────────────────────────
    dyn = np.stack([
        g_hist,
        ind_local,    # global × ind_share — 0 pour nœuds sans conso
        dis_local,    # global × res_share — 0 pour nœuds sans conso
        exp_loc,
        ir24_local,   # rolling 24h × ind_share
        dr24_local,   # rolling 24h × res_share
        g_lag1, g_lag2, g_lag3, g_lag24,
    ], axis=-1).astype(np.float32)   # [T, 28, 10]

    X = np.concatenate([dyn, tf, sf], axis=-1).astype(np.float32)  # [T, 28, 28]
    print(f"\nX_gas_raw : {X.shape}  Y_gas_raw : {g_target.shape}")
    return X, g_target, tgt


X_gas_raw, Y_gas_raw, gas_target_idx0 = build_gas_features()
F_gas = X_gas_raw.shape[-1]
print(f"F_gas = {F_gas}")
 
# --- Features élec ---
def build_elec_features():
    N_E = CFG.num_elec_nodes
    buses = buses_df.copy().sort_values("Node").reset_index(drop=True)
    ls_a = buses["Load_share_%"].to_numpy(dtype=np.float32)/100.
    on_a = buses["Onshore_share_%"].to_numpy(dtype=np.float32)/100.

    ell  = (elec_load_arr[:,None] * ls_a[None,:]).astype(np.float32)
    wof  = np.zeros((T,N_E),dtype=np.float32); wof[:,38] = wind_arr
    won  = (wind_arr[:,None] * on_a[None,:] * 0.2).astype(np.float32)
    pv   = np.zeros((T,N_E),dtype=np.float32)

    lr24 = pd.Series(elec_load_arr).rolling(24,min_periods=1).mean().to_numpy(dtype=np.float32)
    el1  = np.zeros(T,dtype=np.float32); el1[1:]   = elec_load_arr[:-1]
    el24 = np.zeros(T,dtype=np.float32); el24[24:] = elec_load_arr[:-24]
    lr24_r = np.repeat(lr24[:,None],N_E,1).astype(np.float32)
    el1_r  = np.repeat(el1[:,None], N_E,1).astype(np.float32)
    el24_r = np.repeat(el24[:,None],N_E,1).astype(np.float32)

    hour = pd.DatetimeIndex(dt_series).hour
    dow  = pd.DatetimeIndex(dt_series).dayofweek
    tf = np.stack([hour/23., dow/6., (hour==23).astype(float)], -1)
    tf = np.repeat(tf[:,None,:],N_E,1).astype(np.float32)

    sf  = np.repeat(graph_elec["static"][None,:,:],T,0)
    dyn = np.stack([ell,wof,won,pv,lr24_r,el1_r,el24_r],-1).astype(np.float32)
    return np.concatenate([dyn,tf,sf],-1).astype(np.float32)


# ── Définir Y_elec_raw globalement ────────────────────────────────────────
Y_elec_raw = p_raw.to_numpy()[:T].astype(np.float32)   # [T, 24] unités 100 MW

X_elec_raw = build_elec_features()
F_elec     = X_elec_raw.shape[-1]

print(f"X_elec_raw : {X_elec_raw.shape}")
print(f"Y_elec_raw : {Y_elec_raw.shape}  range {Y_elec_raw.min():.2f}..{Y_elec_raw.max():.2f}")
print(f"F_elec = {F_elec}")
```

```python
# ── Vérification des dimensions ───────────────────────────────────────────
F_gas   = X_gas_raw.shape[-1]    # 28
F_elec  = X_elec_raw.shape[-1]   # 20
N_gas   = CFG.num_gas_nodes      # 28
N_elec  = CFG.num_elec_nodes     # 45

train_end = int(T * CFG.train_ratio)
val_end   = int(T * (CFG.train_ratio + CFG.val_ratio))
print(f"F_gas={F_gas}  F_elec={F_elec}  N_gas={N_gas}  N_elec={N_elec}")
print(f"Split : train 0..{train_end}  val {train_end}..{val_end}  test {val_end}..{T}")

# Scalers X
x_gas_scaler  = StandardScaler()
x_elec_scaler = StandardScaler()

X_gas_s = x_gas_scaler.fit_transform(
    X_gas_raw[:train_end].reshape(-1, F_gas)
).reshape(train_end, N_gas, F_gas).astype(np.float32)
X_gas_s = x_gas_scaler.transform(
    X_gas_raw.reshape(-1, F_gas)
).reshape(T, N_gas, F_gas).astype(np.float32)

X_elec_s = x_elec_scaler.fit_transform(
    X_elec_raw[:train_end].reshape(-1, F_elec)
).reshape(train_end, N_elec, F_elec).astype(np.float32)
X_elec_s = x_elec_scaler.transform(
    X_elec_raw.reshape(-1, F_elec)
).reshape(T, N_elec, F_elec).astype(np.float32)

# Scalers Y gaz — 1 par puits
y_gas_scalers = []
for i in range(4):
    sc = StandardScaler(); sc.fit(Y_gas_raw[:train_end, i:i+1])
    y_gas_scalers.append(sc)
Y_gas_s = np.zeros_like(Y_gas_raw, dtype=np.float32)
for i, sc in enumerate(y_gas_scalers):
    Y_gas_s[:,i:i+1] = sc.transform(Y_gas_raw[:,i:i+1])

# Scalers Y élec — 1 par générateur
y_elec_scalers = []
for i in range(CFG.num_generators):
    sc = StandardScaler(); sc.fit(Y_elec_raw[:train_end, i:i+1])
    y_elec_scalers.append(sc)
Y_elec_s = np.zeros_like(Y_elec_raw, dtype=np.float32)
for i, sc in enumerate(y_elec_scalers):
    Y_elec_s[:,i:i+1] = sc.transform(Y_elec_raw[:,i:i+1])

print("Normalisation OK")
print(f"X_gas_s  : {X_gas_s.shape}  X_elec_s : {X_elec_s.shape}")
print(f"Y_gas_s  : {Y_gas_s.shape}  Y_elec_s : {Y_elec_s.shape}")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# FENÊTRAGE
# ════════════════════════════════════════════════════════════════════════════
 
def build_windows_coupled(X_g, X_e, Y_g, Y_e, input_len, output_len, train_end, val_end):
    Xgs, Xes, Ygs, Yes, labels = [], [], [], [], []
    max_s = len(X_g) - input_len - output_len + 1
    for s in range(max_s):
        ie  = s + input_len
        oe  = ie + output_len
        if oe <= train_end:      split = "train"
        elif oe <= val_end:      split = "val"
        else:                    split = "test"
        Xgs.append(X_g[s:ie]); Xes.append(X_e[s:ie])
        Ygs.append(Y_g[ie:oe]); Yes.append(Y_e[ie:oe])
        labels.append(split)
    return (np.stack(Xgs).astype(np.float32),
            np.stack(Xes).astype(np.float32),
            np.stack(Ygs).astype(np.float32),
            np.stack(Yes).astype(np.float32),
            np.array(labels))
 
Xg_seq, Xe_seq, Yg_seq, Ye_seq, split_labels = build_windows_coupled(
    X_gas_s, X_elec_s, Y_gas_s, Y_elec_s,
    CFG.input_len, CFG.output_len, train_end, val_end
)
print(f"\nFenêtrage :")
print(f"  Xg_seq : {Xg_seq.shape}  Xe_seq : {Xe_seq.shape}")
print(f"  Yg_seq : {Yg_seq.shape}  Ye_seq : {Ye_seq.shape}")
print(f"  train={np.sum(split_labels=='train')}  val={np.sum(split_labels=='val')}  test={np.sum(split_labels=='test')}")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# DATASET & DATALOADER
# ════════════════════════════════════════════════════════════════════════════
 
class CoupledDataset(Dataset):
    def __init__(self, Xg, Xe, Yg, Ye):
        self.Xg = torch.tensor(Xg, dtype=torch.float32)
        self.Xe = torch.tensor(Xe, dtype=torch.float32)
        self.Yg = torch.tensor(Yg, dtype=torch.float32)
        self.Ye = torch.tensor(Ye, dtype=torch.float32)
    def __len__(self): return len(self.Xg)
    def __getitem__(self, i): return self.Xg[i], self.Xe[i], self.Yg[i], self.Ye[i]
 
def make_loader(split):
    mask = split_labels == split
    ds   = CoupledDataset(Xg_seq[mask], Xe_seq[mask], Yg_seq[mask], Ye_seq[mask])
    return DataLoader(ds, batch_size=CFG.batch_size,
                      shuffle=(split=="train"),
                      num_workers=0)   # ← remplacer CFG.num_workers par 0

train_loader = make_loader("train")
val_loader   = make_loader("val")
test_loader  = make_loader("test")
print(f"Loaders : train={len(train_loader.dataset)}  val={len(val_loader.dataset)}  test={len(test_loader.dataset)}")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# MODÈLES
# ════════════════════════════════════════════════════════════════════════════
 
class EdgeAwareGraphConv(nn.Module):
    def __init__(self, in_dim, out_dim, edge_dim, edge_hidden=16, dropout=0.0):
        super().__init__()
        self.msg_lin  = nn.Linear(in_dim, out_dim, bias=False)
        self.self_lin = nn.Linear(in_dim, out_dim, bias=True)
        self.edge_mlp = nn.Sequential(nn.Linear(edge_dim,edge_hidden),nn.ReLU(),nn.Linear(edge_hidden,1))
        self.dropout  = nn.Dropout(dropout)
 
    def forward(self, x, edge_index, edge_attr):
        B,N,_ = x.shape
        src,dst = edge_index[0], edge_index[1]
        msg = self.msg_lin(x[:,src,:]) * torch.sigmoid(self.edge_mlp(edge_attr)).unsqueeze(0)
        out = self.self_lin(x)
        agg = torch.zeros(B,N,msg.shape[-1],device=x.device)
        agg.scatter_add_(1, dst.view(1,-1,1).expand(B,-1,msg.shape[-1]), msg)
        deg = torch.zeros(N,device=x.device)
        deg.scatter_add_(0,dst,torch.ones(len(dst),device=x.device))
        return self.dropout(F.relu(out + agg/deg.clamp_min(1.).view(1,N,1)))
 
 
class AdaptiveGraphConv(nn.Module):
    def __init__(self, N, in_dim, out_dim, emb_dim=16, dropout=0.0):
        super().__init__()
        self.node_emb = nn.Parameter(torch.randn(N,emb_dim)*0.01)
        self.msg_lin  = nn.Linear(in_dim,out_dim,bias=False)
        self.self_lin = nn.Linear(in_dim,out_dim,bias=True)
        self.dropout  = nn.Dropout(dropout)
 
    def forward(self, x):
        B,N,_ = x.shape
        A = torch.softmax(F.relu(self.node_emb@self.node_emb.t()),dim=-1)
        return self.dropout(F.relu(self.self_lin(x)+torch.bmm(A.unsqueeze(0).expand(B,-1,-1),self.msg_lin(x))))
 
 
class CoupledEnergyGNN(nn.Module):
    def __init__(self, input_len, F_gas, F_elec, edge_dim_gas, edge_dim_elec,
                 hidden_dim, gnn_hidden, edge_hidden, emb_dim, output_len,
                 gas_target_idx0, gen_bus_idx, dropout=0.15):
        super().__init__()
        self.output_len      = output_len
        self.gas_target_idx0 = gas_target_idx0
        self.gen_bus_idx     = gen_bus_idx
        fg, fe = input_len*F_gas, input_len*F_elec
 
        def enc(flat):
            return nn.Sequential(
                nn.Linear(flat,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout),
                nn.Linear(hidden_dim,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout))
 
        self.encoder_gas  = enc(fg)
        self.encoder_elec = enc(fe)
        self.gconv_gas1  = EdgeAwareGraphConv(hidden_dim,gnn_hidden,edge_dim_gas,edge_hidden,dropout)
        self.gconv_gas2  = EdgeAwareGraphConv(gnn_hidden,gnn_hidden,edge_dim_gas,edge_hidden,dropout)
        self.adapt_gas   = AdaptiveGraphConv(28,gnn_hidden,gnn_hidden,emb_dim,dropout)
        self.norm_gas    = nn.LayerNorm(gnn_hidden)
        self.gconv_elec1 = EdgeAwareGraphConv(hidden_dim,gnn_hidden,edge_dim_elec,edge_hidden,dropout)
        self.gconv_elec2 = EdgeAwareGraphConv(gnn_hidden,gnn_hidden,edge_dim_elec,edge_hidden,dropout)
        self.adapt_elec  = AdaptiveGraphConv(45,gnn_hidden,gnn_hidden,emb_dim,dropout)
        self.norm_elec   = nn.LayerNorm(gnn_hidden)
        self.mlp_g2e     = nn.Sequential(nn.Linear(gnn_hidden,gnn_hidden),nn.ReLU())
        self.gate_g2e    = nn.Linear(gnn_hidden*2,gnn_hidden)
        self.mlp_e2g     = nn.Sequential(nn.Linear(gnn_hidden,gnn_hidden),nn.ReLU())
        self.gate_e2g    = nn.Linear(gnn_hidden*2,gnn_hidden)
        self.norm_pg     = nn.LayerNorm(gnn_hidden)
        self.norm_pe     = nn.LayerNorm(gnn_hidden)
 
        def dec():
            return nn.Sequential(
                nn.Linear(gnn_hidden,gnn_hidden),nn.LayerNorm(gnn_hidden),
                nn.ReLU(),nn.Dropout(dropout),nn.Linear(gnn_hidden,output_len))
 
        self.decoder_gas  = dec()
        self.decoder_elec = dec()
 
    def forward(self, xg, xe, ei_g, ea_g, ei_e, ea_e,
            ci_g, ci_e, c_eff, c_pmax):
        B,Tin,Ng,Fg = xg.shape
        _,_,Ne,Fe   = xe.shape

        # ── Étape 1 : Encodage ───────────────────────────────────────────────
        h_g = self.encoder_gas( xg.permute(0,2,1,3).reshape(B*Ng,Tin*Fg)).view(B,Ng,-1)
        h_e = self.encoder_elec(xe.permute(0,2,1,3).reshape(B*Ne,Tin*Fe)).view(B,Ne,-1)

    # ── Étape 2 : Propagation intra-graphe ──────────────────────────────
    # ── ORIGINAL (erreur) ─────────────────────────────────────────────────
    # h_g = self.norm_gas(h_g + self.adapt_gas(self.gconv_gas2(...)))
    # ← h_g=[B,28,128] mais gconv_gas1 sort [B,28,64] → dim mismatch
    # ── CORRIGÉ ───────────────────────────────────────────────────────────
        h_g_conv = self.gconv_gas2(self.gconv_gas1(h_g, ei_g, ea_g), ei_g, ea_g)
        h_g = self.norm_gas(h_g_conv + self.adapt_gas(h_g_conv))

        h_e_conv = self.gconv_elec2(self.gconv_elec1(h_e, ei_e, ea_e), ei_e, ea_e)
        h_e = self.norm_elec(h_e_conv + self.adapt_elec(h_e_conv))

    # ── Étape 3 : Couplage bidirectionnel ────────────────────────────────
        D   = h_g.shape[-1]
        hgc = h_g[:,ci_g,:]; hec = h_e[:,ci_e,:]

        msg_g2e = self.mlp_g2e(hgc) * c_eff.view(1,-1,1)
        dlt_e   = torch.sigmoid(self.gate_g2e(torch.cat([hec,msg_g2e],-1))) * msg_g2e

        msg_e2g = self.mlp_e2g(hec) * c_pmax.view(1,-1,1)
        dlt_g   = torch.sigmoid(self.gate_e2g(torch.cat([hgc,msg_e2g],-1))) * msg_e2g

        h_g = self.norm_pg(h_g.clone().scatter_add_(
            1, ci_g.view(1,-1,1).expand(B,-1,D), dlt_g))
        h_e = self.norm_pe(h_e.clone().scatter_add_(
            1, ci_e.view(1,-1,1).expand(B,-1,D), dlt_e))

    # ── Étape 4 : Décodage ───────────────────────────────────────────────
        y_g = self.decoder_gas( h_g[:,self.gas_target_idx0,:]).permute(0,2,1).contiguous()
        y_e = self.decoder_elec(h_e[:,self.gen_bus_idx,:]    ).permute(0,2,1).contiguous()
        return y_g, y_e
 
 
class MLPBaselineGas(nn.Module):
    def __init__(self,input_len,F_gas,hidden_dim,output_len,tgt_idx,dropout=0.15):
        super().__init__()
        self.tgt=tgt_idx; flat=input_len*F_gas
        self.enc=nn.Sequential(nn.Linear(flat,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout),
                               nn.Linear(hidden_dim,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout))
        self.dec=nn.Sequential(nn.Linear(hidden_dim,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout),nn.Linear(hidden_dim,output_len))
    def forward(self,xg):
        B,T,N,F=xg.shape; h=self.enc(xg.permute(0,2,1,3).reshape(B*N,T*F)).view(B,N,-1)
        return self.dec(h[:,self.tgt,:]).permute(0,2,1).contiguous()
 
 
class MLPBaselineElec(nn.Module):
    def __init__(self,input_len,F_elec,hidden_dim,output_len,gen_bus,dropout=0.15):
        super().__init__()
        self.gb=gen_bus; flat=input_len*F_elec
        self.enc=nn.Sequential(nn.Linear(flat,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout),
                               nn.Linear(hidden_dim,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout))
        self.dec=nn.Sequential(nn.Linear(hidden_dim,hidden_dim),nn.LayerNorm(hidden_dim),nn.ReLU(),nn.Dropout(dropout),nn.Linear(hidden_dim,output_len))
    def forward(self,xe):
        B,T,N,F=xe.shape; h=self.enc(xe.permute(0,2,1,3).reshape(B*N,T*F)).view(B,N,-1)
        return self.dec(h[:,self.gb,:]).permute(0,2,1).contiguous()
```

```python
# ════════════════════════════════════════════════════════════════════════════
# TENSEURS GRAPHES & COUPLAGE
# ════════════════════════════════════════════════════════════════════════════
 
# ── Extraire les arrays depuis les dicts ──────────────────────────────────
edge_index_gas  = graph_gas["edge_index"]
edge_attr_gas   = graph_gas["edge_attr"]
edge_index_elec = graph_elec["edge_index"]
edge_attr_elec  = graph_elec["edge_attr"]

# ── Tenseurs sur device ───────────────────────────────────────────────────
ei_g  = torch.tensor(edge_index_gas,  dtype=torch.long,    device=device)
ea_g  = torch.tensor(edge_attr_gas,   dtype=torch.float32, device=device)
ei_e  = torch.tensor(edge_index_elec, dtype=torch.long,    device=device)
ea_e  = torch.tensor(edge_attr_elec,  dtype=torch.float32, device=device)
ci_g  = torch.tensor(coup_gas_idx,    dtype=torch.long,    device=device)
ci_e  = torch.tensor(coup_elec_idx,   dtype=torch.long,    device=device)
c_eff = torch.tensor(coup_eff,        dtype=torch.float32, device=device)
c_pmx = torch.tensor(coup_pmax_norm,  dtype=torch.float32, device=device)
gen_bus_t = torch.tensor(gen_bus_idx, dtype=torch.long,    device=device)

print(f"ei_g={ei_g.shape}  ea_g={ea_g.shape}")
print(f"ei_e={ei_e.shape}  ea_e={ea_e.shape}")
print(f"coupling gas:{ci_g.shape}  elec:{ci_e.shape}")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# ENTRAÎNEMENT
# ════════════════════════════════════════════════════════════════════════════
 
class EarlyStopping:
    def __init__(self, patience=15):
        self.patience=patience; self.best=np.inf; self.bad=0
    def step(self, v):
        if v < self.best:
            self.best=v; self.bad=0; return True, False
        self.bad+=1
        return False, self.bad>=self.patience
 
def inverse_transform_y(y_scaled, scalers):
    y=np.empty_like(y_scaled,dtype=np.float32)
    for j,sc in enumerate(scalers):
        y[...,j]=sc.inverse_transform(y_scaled[...,j].reshape(-1,1)).reshape(y_scaled.shape[:-1])
    return y
 
def run_epoch(model, loader, optimizer=None, mode="coupled"):
    is_train = optimizer is not None
    model.train(is_train)
    criterion = nn.MSELoss()
    total_loss=0.; yg_list,ye_list,tg_list,te_list=[],[],[],[]
 
    for xg,xe,yg,ye in loader:
        xg,xe,yg,ye = xg.to(device),xe.to(device),yg.to(device),ye.to(device)
        if is_train: optimizer.zero_grad()
 
        if mode == "coupled":
            pg, pe = model(xg,xe,ei_g,ea_g,ei_e,ea_e,ci_g,ci_e,c_eff,c_pmx)
            loss = CFG.alpha*criterion(pg,yg) + (1-CFG.alpha)*criterion(pe,ye)
        elif mode == "gas":
            pg = model(xg); pe = yg*0; loss = criterion(pg, yg)
        else:
            pe = model(xe); pg = ye*0; loss = criterion(pe, ye)
 
        if is_train:
            loss.backward()
            if CFG.grad_clip: torch.nn.utils.clip_grad_norm_(model.parameters(), CFG.grad_clip)
            optimizer.step()
 
        total_loss += loss.item() * xg.size(0)
        yg_list.append(pg.detach().cpu().numpy())
        ye_list.append(pe.detach().cpu().numpy())
        tg_list.append(yg.detach().cpu().numpy())
        te_list.append(ye.detach().cpu().numpy())
 
    avg = total_loss / len(loader.dataset)
    return avg, np.concatenate(yg_list), np.concatenate(ye_list), \
               np.concatenate(tg_list), np.concatenate(te_list)
 
 
def fit_model(model, name="model", mode="coupled"):
    model = model.to(device)
    opt   = torch.optim.AdamW(model.parameters(), lr=CFG.lr, weight_decay=CFG.weight_decay)
    sch   = torch.optim.lr_scheduler.ReduceLROnPlateau(opt, mode="min", factor=0.5, patience=8, min_lr=1e-6)
    early = EarlyStopping(CFG.patience)
    hist  = {"train":[], "val":[]}
    path  = os.path.join(CFG.output_dir, f"best_{name}.pt")
 
    for ep in range(1, CFG.max_epochs+1):
        tl,_,_,_,_ = run_epoch(model, train_loader, opt,  mode)
        vl,_,_,_,_ = run_epoch(model, val_loader,   None, mode)
        hist["train"].append(tl); hist["val"].append(vl)
        sch.step(vl)
        improved, stop = early.step(vl)
        if improved: torch.save(model.state_dict(), path)
        print(f"Ep {ep:03d} | train={tl:.5f} | val={vl:.5f} | best={early.best:.5f} | lr={opt.param_groups[0]['lr']:.2e}")
        if stop: print("Early stopping."); break
 
    model.load_state_dict(torch.load(path, map_location=device))
    return model, hist
 
 
def predict_test(model, mode="coupled"):
    _,pg,pe,tg,te = run_epoch(model, test_loader, None, mode)

    pg_d = inverse_transform_y(pg, y_gas_scalers)
    tg_d = inverse_transform_y(tg, y_gas_scalers)

    # ── ORIGINAL ──────────────────────────────────────────────────────────
    # pe_d = inverse_transform_y(pe, y_elec_scalers)  ← échoue si mode="gas"
    # ── CORRIGÉ : inverse_transform seulement si les dims correspondent ───
    if mode == "gas":
        pe_d = pe   # déjà à zéro, pas besoin de dénormaliser
        te_d = te
    else:
        pe_d = inverse_transform_y(pe, y_elec_scalers)
        te_d = inverse_transform_y(te, y_elec_scalers)

    return pg_d, pe_d, tg_d, te_d
 
 
def compute_metrics(y_true, y_pred, name=""):
    err=y_pred-y_true; mae=np.mean(np.abs(err)); rmse=np.sqrt(np.mean(err**2))
    mean_abs=np.mean(np.abs(y_true))
    mae_mean = mae/mean_abs*100 if mean_abs>0 else np.nan
    ss_res=np.sum((y_true-y_pred)**2); ss_tot=np.sum((y_true-y_true.mean())**2)
    r2=float(1-ss_res/ss_tot) if ss_tot>0 else np.nan
    print(f"{name}  MAE={mae:.4f}  RMSE={rmse:.4f}  MAE/MEAN={mae_mean:.2f}%  R²={r2:.4f}")
    return {"MAE":mae,"RMSE":rmse,"MAE/MEAN":mae_mean,"R2":r2}
```

```python
# ════════════════════════════════════════════════════════════════════════════
# INSTANCIATION & ENTRAÎNEMENT
# ════════════════════════════════════════════════════════════════════════════
 
print("\n=== Instanciation des modèles ===")

# ── Variables manquantes à définir globalement ────────────────────────────
gas_target_idx0 = [n-1 for n in CFG.gas_well_nodes_1based]  # [0,4,16,17]
F_gas           = X_gas_raw.shape[-1]
F_elec          = X_elec_raw.shape[-1]

coupled_model = CoupledEnergyGNN(
    input_len=CFG.input_len, F_gas=F_gas, F_elec=F_elec,
    edge_dim_gas=6, edge_dim_elec=4,
    hidden_dim=CFG.hidden_dim, gnn_hidden=CFG.gnn_hidden,
    edge_hidden=CFG.edge_hidden, emb_dim=CFG.emb_dim,
    output_len=CFG.output_len,
    gas_target_idx0=gas_target_idx0,
    gen_bus_idx=gen_bus_t,
    dropout=CFG.dropout,
)
 
baseline_gas  = MLPBaselineGas( CFG.input_len, F_gas,  CFG.hidden_dim, CFG.output_len, gas_target_idx0, CFG.dropout)
baseline_elec = MLPBaselineElec(CFG.input_len, F_elec, CFG.hidden_dim, CFG.output_len, gen_bus_t,        CFG.dropout)
 
def count_params(m):
    n=sum(p.numel() for p in m.parameters() if p.requires_grad)
    print(f"  {m.__class__.__name__:<25s} : {n:,} params")
    return n
 
print("Paramètres :")
count_params(coupled_model)
count_params(baseline_gas)
count_params(baseline_elec)
 
# Test forward
coupled_model.to(device)
coupled_model.eval()
with torch.no_grad():
    xg_t, xe_t, yg_t, ye_t = next(iter(train_loader))
    out_g, out_e = coupled_model(
        xg_t.to(device), xe_t.to(device),
        ei_g, ea_g, ei_e, ea_e, ci_g, ci_e, c_eff, c_pmx
    )
print(f"\nTest forward OK : y_gas={out_g.shape}  y_elec={out_e.shape}")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# ENTRAÎNEMENT
# ════════════════════════════════════════════════════════════════════════════
 
print("\n" + "="*60)
print("ENTRAÎNEMENT — Coupled GNN")
print("="*60)
coupled_model, hist_coupled = fit_model(coupled_model, "coupled_gnn", mode="coupled")
 
print("\n" + "="*60)
print("ENTRAÎNEMENT — Baseline Gaz")
print("="*60)
baseline_gas, hist_base_g = fit_model(baseline_gas, "baseline_gas", mode="gas")
 
print("\n" + "="*60)
print("ENTRAÎNEMENT — Baseline Élec")
print("="*60)
baseline_elec, hist_base_e = fit_model(baseline_elec, "baseline_elec", mode="elec")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# ÉVALUATION
# ════════════════════════════════════════════════════════════════════════════
 
print("\n=== ÉVALUATION SUR TEST SET ===")
 
pg_c, pe_c, tg_c, te_c = predict_test(coupled_model, "coupled")
pg_b, _,    tg_b, _    = predict_test(baseline_gas,  "gas")
_,    pe_b, _,    te_b = predict_test(baseline_elec, "elec")
 
print("\n--- GAZ (injection puits) ---")
m_cg = compute_metrics(tg_c, pg_c, "Coupled GNN  ")
m_bg = compute_metrics(tg_b, pg_b, "Baseline Gas ")
 
print("\n--- ÉLECTRICITÉ (production générateurs) ---")
m_ce = compute_metrics(te_c, pe_c, "Coupled GNN  ")
m_be = compute_metrics(te_b, pe_b, "Baseline Elec")
 
print("\n--- GAZ agrégé (somme 4 puits) ---")
total_g_true  = tg_c.sum(-1); total_g_c = pg_c.sum(-1); total_g_b = pg_b.sum(-1)
m_agg_gc = compute_metrics(total_g_true, total_g_c, "Coupled agg-gaz")
m_agg_gb = compute_metrics(total_g_true, total_g_b, "Baseline agg-gaz")
 
print("\n--- ÉLEC agrégé (somme 24 générateurs) ---")
total_e_true  = te_c.sum(-1); total_e_c = pe_c.sum(-1); total_e_b = pe_b.sum(-1)
m_agg_ec = compute_metrics(total_e_true, total_e_c, "Coupled agg-elec")
m_agg_eb = compute_metrics(total_e_true, total_e_b, "Baseline agg-elec")
 
# MAE/MEAN par horizon
def mae_per_horizon(y_true, y_pred):
    err = np.abs(y_pred - y_true)
    mean_h = np.mean(np.abs(y_true), axis=(0,2))
    mae_h  = np.mean(err, axis=(0,2))
    return mae_h / np.clip(mean_h, 1e-8, None) * 100
 
print("\n--- MAE/MEAN par horizon (gaz) ---")
for h, (vc, vb) in enumerate(zip(mae_per_horizon(tg_c, pg_c), mae_per_horizon(tg_b, pg_b)), 1):
    print(f"  h+{h:02d} : Coupled={vc:.2f}%  Baseline={vb:.2f}%")
 
print("\n--- MAE/MEAN par horizon (élec) ---")
for h, (vc, vb) in enumerate(zip(mae_per_horizon(te_c, pe_c), mae_per_horizon(te_b, pe_b)), 1):
    print(f"  h+{h:02d} : Coupled={vc:.2f}%  Baseline={vb:.2f}%")
```

```python
# ════════════════════════════════════════════════════════════════════════════
# VISUALISATIONS
# ════════════════════════════════════════════════════════════════════════════
 
print("\n=== Génération des plots ===")
 
node_names_g = [f"Well {n}" for n in CFG.gas_well_nodes_1based]
gen_names    = gen_df["Name"].tolist()
 
# 1. Courbes de loss
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for ax, hist, title in [
    (axes[0], {"GNN train": hist_coupled["train"], "GNN val": hist_coupled["val"],
               "Base-G train": hist_base_g["train"], "Base-G val": hist_base_g["val"]}, "Loss gaz+élec"),
    (axes[1], {"GNN train": hist_coupled["train"], "GNN val": hist_coupled["val"],
               "Base-E train": hist_base_e["train"], "Base-E val": hist_base_e["val"]}, "Loss électricité"),
]:
    for k, v in hist.items():
        ls = "--" if "Base" in k else "-"
        ax.plot(v, label=k, linestyle=ls)
    ax.set_xlabel("Epoch"); ax.set_ylabel("Loss"); ax.set_title(title)
    ax.legend(fontsize=8); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(CFG.output_dir, "01_loss_curves.png"), dpi=100)
plt.show(); print("Plot 1 sauvegardé")
 
# 2. Prédictions gaz — one-step
n_plot = min(72, len(pg_c))
fig, axes = plt.subplots(4, 1, figsize=(12, 10), sharex=True)
for i, ax in enumerate(axes):
    ax.plot(tg_c[:n_plot, 0, i],  label="True", color="steelblue", lw=1.5)
    ax.plot(pg_c[:n_plot, 0, i],  label="Coupled GNN", color="darkorange", lw=1.5)
    ax.plot(pg_b[:n_plot, 0, i],  label="Baseline", color="green", lw=1, ls="--")
    ax.set_ylabel(node_names_g[i]); ax.grid(alpha=0.3)
axes[0].legend(ncol=3, fontsize=9)
axes[-1].set_xlabel("Test samples")
fig.suptitle("Prédiction h+1 — Injection gaz (MNm3/h)")
plt.tight_layout()
plt.savefig(os.path.join(CFG.output_dir, "02_gas_predictions.png"), dpi=100)
plt.show(); print("Plot 2 sauvegardé")
 
# 3. Prédictions élec — agrégé
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(total_e_true[:n_plot, 0],   label="True total", color="steelblue", lw=1.5)
ax.plot(total_e_c[:n_plot, 0],      label="Coupled GNN total", color="darkorange", lw=1.5)
ax.plot(total_e_b[:n_plot, 0],      label="Baseline total", color="green", lw=1, ls="--")
ax.set_xlabel("Test samples"); ax.set_ylabel("Production totale [MW]")
ax.set_title("Production électrique totale — h+1")
ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(CFG.output_dir, "03_elec_aggregate.png"), dpi=100)
plt.show(); print("Plot 3 sauvegardé")
 
# 4. MAE/MEAN par horizon (les deux sous-systèmes)
mh_gc = mae_per_horizon(tg_c, pg_c)
mh_gb = mae_per_horizon(tg_b, pg_b)
mh_ec = mae_per_horizon(te_c, pe_c)
mh_eb = mae_per_horizon(te_b, pe_b)
hs    = np.arange(1, CFG.output_len+1)
 
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].plot(hs, mh_gc, "o-", label="Coupled GNN", color="darkorange")
axes[0].plot(hs, mh_gb, "s--", label="Baseline", color="steelblue")
axes[0].set_xlabel("Horizon (h)"); axes[0].set_ylabel("MAE/MEAN [%]")
axes[0].set_title("Gaz — MAE/MEAN par horizon"); axes[0].legend(); axes[0].grid(alpha=0.3)
 
axes[1].plot(hs, mh_ec, "o-", label="Coupled GNN", color="darkorange")
axes[1].plot(hs, mh_eb, "s--", label="Baseline", color="steelblue")
axes[1].set_xlabel("Horizon (h)"); axes[1].set_ylabel("MAE/MEAN [%]")
axes[1].set_title("Élec — MAE/MEAN par horizon"); axes[1].legend(); axes[1].grid(alpha=0.3)
 
plt.tight_layout()
plt.savefig(os.path.join(CFG.output_dir, "04_mae_by_horizon.png"), dpi=100)
plt.show(); print("Plot 4 sauvegardé")
 
# 5. Heatmap gaz [4 puits × 24h]
rmse_gas_hmap = np.sqrt(np.mean((pg_c - tg_c)**2, axis=0)).T   # [4, 24]
fig, ax = plt.subplots(figsize=(12, 3))
im = ax.imshow(rmse_gas_hmap, aspect="auto", cmap="YlOrRd")
plt.colorbar(im, ax=ax, label="RMSE [MNm3/h]")
ax.set_yticks(range(4)); ax.set_yticklabels(node_names_g)
ax.set_xticks(range(CFG.output_len)); ax.set_xticklabels(range(1, CFG.output_len+1))
ax.set_xlabel("Horizon (h)"); ax.set_title("Heatmap RMSE gaz — puits × horizon")
plt.tight_layout()
plt.savefig(os.path.join(CFG.output_dir, "05_heatmap_gas.png"), dpi=100)
plt.show(); print("Plot 5 sauvegardé")
 
# 6. Heatmap élec [24 générateurs × 24h]
rmse_elec_hmap = np.sqrt(np.mean((pe_c - te_c)**2, axis=0)).T   # [24, 24]
fig, ax = plt.subplots(figsize=(14, 8))
im = ax.imshow(rmse_elec_hmap, aspect="auto", cmap="YlOrRd")
plt.colorbar(im, ax=ax, label="RMSE [MW]")
ax.set_yticks(range(CFG.num_generators))
ax.set_yticklabels([f"G{i+1}" for i in range(CFG.num_generators)], fontsize=8)
ax.set_xticks(range(CFG.output_len)); ax.set_xticklabels(range(1, CFG.output_len+1), fontsize=8)
ax.set_xlabel("Horizon (h)"); ax.set_title("Heatmap RMSE élec — générateur × horizon")
plt.tight_layout()
plt.savefig(os.path.join(CFG.output_dir, "06_heatmap_elec.png"), dpi=100)
plt.show(); print("Plot 6 sauvegardé")
```
