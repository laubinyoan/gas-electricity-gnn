```python
import numpy as np                                  # numerical arrays
import pandas as pd                                 # excel / dataframe loading
import torch                                        # main deep learning framework
import torch.nn as nn                              # neural network building blocks
import torch.nn.functional as F                    # activation functions, etc.
from torch.utils.data import Dataset, DataLoader   # batching utilities
from torch_geometric.nn import MessagePassing       # pyg base class for graph convolutions
from sklearn.preprocessing import StandardScaler   # z-score normalisation
import matplotlib.pyplot as plt                    # plots

torch.manual_seed(42)   # fix random seed for reproducibility
np.random.seed(42)      # fix numpy seed as well
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')   # use gpu if available
print('Device:', device)
```

```python
nodes_df = pd.read_excel('./Gas Network Data.xlsx', sheet_name='Nodes')       # node metadata (shares, etc.)
pipes_df = pd.read_excel('./Gas Network Data.xlsx', sheet_name='Pipelines')    # pipeline topology

months = [str(m).zfill(2) + '-2024' for m in range(1, 13)]   # list of sheet names for each month

res_total = pd.concat(                                                          # residential consumption [T]
    [pd.read_excel('./input/Gas Cons Distrib.xlsx', sheet_name=m)['Physical Flow (MNm3)'] for m in months],
    ignore_index=True
).values.astype(np.float32)

ind_total = pd.concat(                                                          # industrial consumption [T]
    [pd.read_excel('./input/Gas Cons Ind.xlsx', sheet_name=m)['Physical Flow (MNm3)'] for m in months],
    ignore_index=True
).values.astype(np.float32)

exp_total = pd.concat(                                                          # export consumption [T]
    [pd.read_excel('./input/Gas Cons Export.xlsx', sheet_name=m)['Physical Flow (MNm3)'] for m in months],
    ignore_index=True
).values.astype(np.float32)

q_full = pd.concat(                                                             # pipeline flow rates [T, 33]
    [pd.read_excel('./output/Decision_variablesRU.xlsx', sheet_name=f'q-{m}').iloc[:, 1:] for m in months],
    ignore_index=True
).values.astype(np.float32)

g_full = pd.concat(                                                             # well injections [T, 4] — this is the target
    [pd.read_excel('./output/Decision_variablesRU.xlsx', sheet_name=f'g-{m}')[[0, 1, 2, 3]] for m in months],
    ignore_index=True
).values.astype(np.float32)

print('q_full:', q_full.shape)
print('g_full:', g_full.shape)
```

```python
N        = 28       # number of nodes in the gas network
T        = 8784     # total hours in 2024 (leap year)
W_IN     = 72       # input window : past 72 hours
W_OUT    = 24       # output window : next 24 hours
BLOCK    = W_IN + W_OUT
WELL_IDX = [0, 4, 16, 17]   # 0-based indices of the 4 well nodes

nodes_df  = nodes_df.sort_values('Gas Node').reset_index(drop=True)   # sort nodes by id

res_share = nodes_df['Res_Load_share_%'].values.astype(np.float32)    # fraction of residential conso per node
ind_share = nodes_df['Ind_Load_share_%'].values.astype(np.float32)    # fraction of industrial conso per node
exp_share = nodes_df['Exp_Load_share_%'].values.astype(np.float32)    # fraction of export conso per node

# outer product: [T] x [N] → [T, N]  — assigns each node its share of total consumption
X_cons = np.stack([
    np.outer(res_total, res_share),   # residential demand per node [T, N]
    np.outer(ind_total, ind_share),   # industrial demand per node  [T, N]
    np.outer(exp_total, exp_share),   # export demand per node      [T, N]
], axis=-1).astype(np.float32)        # final shape: [T, N, 3]

# global totals broadcast to all nodes (give each node the network-wide signal)
X_global = np.stack([
    np.tile(res_total[:, None], (1, N)),   # [T, N] same value for all nodes
    np.tile(ind_total[:, None], (1, N)),   # [T, N]
    np.tile(exp_total[:, None], (1, N)),   # [T, N]
], axis=-1).astype(np.float32)

hours    = np.arange(T)
hour_sin = np.sin(2 * np.pi * (hours % 24)      / 24).astype(np.float32)   # cyclic hour of day (sin)
hour_cos = np.cos(2 * np.pi * (hours % 24)      / 24).astype(np.float32)   # cyclic hour of day (cos)
dow_sin  = np.sin(2 * np.pi * (hours // 24 % 7) / 7 ).astype(np.float32)   # cyclic day of week (sin)
dow_cos  = np.cos(2 * np.pi * (hours // 24 % 7) / 7 ).astype(np.float32)   # cyclic day of week (cos)

time_feats = np.stack([                                  # broadcast time features to all nodes
    np.tile(hour_sin[:, None], (1, N)),
    np.tile(hour_cos[:, None], (1, N)),
    np.tile(dow_sin[:,  None], (1, N)),
    np.tile(dow_cos[:,  None], (1, N)),
], axis=-1).astype(np.float32)                           # [T, N, 4]

g_node = np.zeros((T, N), dtype=np.float32)             # placeholder for well injection at node level
g_node[:, WELL_IDX] = g_full                            # only the 4 well nodes have a non-zero injection signal

X_full = np.concatenate([X_cons, X_global, time_feats, g_node[:, :, None]], axis=-1)   # [T, N, 11]

cols_to_norm = list(range(0, 6)) + [10]   # normalise consumption columns + well injection
x_scaler = StandardScaler().fit(X_full[:, :, cols_to_norm].reshape(-1, len(cols_to_norm)))
X_full[:, :, cols_to_norm] = x_scaler.transform(
    X_full[:, :, cols_to_norm].reshape(-1, len(cols_to_norm))
).reshape(T, N, len(cols_to_norm))                      # apply z-score in-place

Y_all    = g_full.astype(np.float32)                    # targets: well injections [T, 4]
y_scaler = StandardScaler().fit(Y_all)                  # fit target scaler on all data
Y_norm   = y_scaler.transform(Y_all).astype(np.float32) # normalised targets

print(f'X_full : {X_full.shape}  (T x N x 11 features)')
```

```python
pipes_df = pipes_df[['Pipeline', 'From node', 'To node',
                      'Compressor factor', 'Estimated lenght [km]']].copy()
pipes_df['Estimated lenght [km]'] = pipes_df['Estimated lenght [km]'].fillna(0.0)   # fill missing lengths
pipes_df['Compressor factor']     = pipes_df['Compressor factor'].fillna(1.0)       # fill missing compressor factors

edge_pairs   = []   # list of (u, v) directed pairs
pipe_to_edge = []   # maps pipeline index to (forward_edge_idx, backward_edge_idx)

for i, row in pipes_df.iterrows():
    u = int(row['From node']) - 1   # convert 1-based node id to 0-based
    v = int(row['To node'])   - 1
    idx_fwd = len(edge_pairs)       # index of the forward edge
    edge_pairs.append((u, v))
    idx_bwd = len(edge_pairs)       # index of the backward edge
    edge_pairs.append((v, u))
    pipe_to_edge.append((idx_fwd, idx_bwd))

edge_index = np.array(edge_pairs, dtype=np.int64).T   # [2, E=66] — pyg standard format
E          = edge_index.shape[1]                       # 66 directed edges

static_edge = np.array([                               # static edge attributes: compressor + length
    [pipes_df.iloc[p]['Compressor factor'],
     pipes_df.iloc[p]['Estimated lenght [km]']]
    for p, _ in enumerate(pipe_to_edge)
    for _ in range(2)                                  # duplicate for fwd and bwd edge
], dtype=np.float32)                                   # [E, 2]

static_scaler    = StandardScaler().fit(static_edge)
static_edge_norm = static_scaler.transform(static_edge).astype(np.float32)   # normalised static attrs

Q_edge = np.zeros((T, E), dtype=np.float32)            # dynamic edge feature: pipeline flow [T, E]
for p, (idx_f, idx_b) in enumerate(pipe_to_edge):
    Q_edge[:, idx_f] = np.abs(q_full[:, p])            # forward edge gets |q|
    Q_edge[:, idx_b] = np.abs(q_full[:, p])            # backward edge same value (undirected flow)

q_scaler    = StandardScaler().fit(Q_edge.reshape(-1, 1))
Q_edge_norm = q_scaler.transform(Q_edge.reshape(-1, 1)).reshape(T, E).astype(np.float32)

# keep only the dynamic flow q as edge feature (1 feature per edge per timestep)
EA_full = Q_edge_norm[:, :, None]   # [T, E, 1]

ei = torch.tensor(edge_index, dtype=torch.long, device=device)   # edge_index on device for pyg

print(f'edge_index : {edge_index.shape}')
print(f'EA_full    : {EA_full.shape}  (T x {E} edges x 1 feature)')
```

```python
starts = np.arange(0, T - BLOCK + 1, 6)   # start index of each window (step=6h)

X_win  = np.stack([X_full[s : s + W_IN]         for s in starts])   # [S, 72, N, 11]
EA_win = np.stack([EA_full[s : s + W_IN]         for s in starts])   # [S, 72, E, 1]
Y_win  = np.stack([Y_norm[s + W_IN : s + BLOCK]  for s in starts])   # [S, 24, 4]

S       = len(starts)
s_train = int(0.70 * S)   # 70% training
s_val   = int(0.85 * S)   # 15% validation

X_train,  Y_train  = X_win[:s_train],       Y_win[:s_train]
X_val,    Y_val    = X_win[s_train:s_val],  Y_win[s_train:s_val]
X_test,   Y_test   = X_win[s_val:],         Y_win[s_val:]
EA_train = EA_win[:s_train]
EA_val   = EA_win[s_train:s_val]
EA_test  = EA_win[s_val:]

print(f'Train: {len(X_train)}  Val: {len(X_val)}  Test: {len(X_test)}')
```

```python
class GasDataset(Dataset):
    def __init__(self, X, EA, Y):
        self.X  = torch.tensor(X,  dtype=torch.float32)   # node features [S, 72, N, 11]
        self.EA = torch.tensor(EA, dtype=torch.float32)   # edge features [S, 72, E, 1]
        self.Y  = torch.tensor(Y,  dtype=torch.float32)   # targets        [S, 24, 4]

    def __len__(self):
        return len(self.X)   # number of windows

    def __getitem__(self, idx):
        return self.X[idx], self.EA[idx], self.Y[idx]   # one sample

BATCH = 32

train_loader = DataLoader(GasDataset(X_train, EA_train, Y_train), batch_size=BATCH, shuffle=True)
val_loader   = DataLoader(GasDataset(X_val,   EA_val,   Y_val),   batch_size=BATCH, shuffle=False)
test_loader  = DataLoader(GasDataset(X_test,  EA_test,  Y_test),  batch_size=BATCH, shuffle=False)

print(f'Batches — train:{len(train_loader)}  val:{len(val_loader)}  test:{len(test_loader)}')
```

```python
H       = 96      # hidden dimension for all layers
IN_DIM  = 11      # number of input features per node
EDGE_DIM = 1      # number of edge features
DROPOUT = 0.2     # dropout probability


def make_batch_edge_index(edge_index, B, N):
    # repeat edge_index B times, adding node offset so each graph in the batch is separate
    offsets = torch.arange(B, device=edge_index.device) * N   # [B] offset per graph
    ei = edge_index.unsqueeze(0).expand(B, -1, -1)            # [B, 2, E]
    return (ei + offsets.view(B, 1, 1)).reshape(2, -1)        # [2, B*E]


class EdgeWeightedConv(MessagePassing):
    # one graph convolution step with learnable edge-weight gating
    def __init__(self, h_dim, edge_dim=1, edge_h=16, dropout=0.2):
        super().__init__(aggr='mean')                                          # average messages from neighbours
        self.edge_mlp = nn.Sequential(                                         # maps edge feature to scalar gate
            nn.Linear(edge_dim, edge_h), nn.ReLU(), nn.Linear(edge_h, 1)
        )
        self.W    = nn.Linear(h_dim, h_dim)    # linear transform applied after aggregation
        self.norm = nn.LayerNorm(h_dim)        # layer norm for training stability
        self.drop = nn.Dropout(dropout)        # dropout for regularisation

    def forward(self, h, edge_index, ea):
        # h          : [B*N, h_dim]  node embeddings (all graphs in batch stacked)
        # edge_index : [2, B*E]      batched directed edges
        # ea         : [B*E, 1]      edge flow feature
        return self.propagate(edge_index, x=h, ea=ea)   # pyg triggers message → aggregate → update

    def message(self, x_j, ea):
        # x_j : [B*E, h_dim]  source node feature for each edge
        # ea  : [B*E, 1]      edge feature for each edge
        w = torch.sigmoid(self.edge_mlp(ea))   # scalar gate in [0,1] per edge
        return x_j * w                         # weighted message from source node

    def update(self, aggr_out, x):
        # aggr_out : [B*N, h_dim]  mean of incoming messages
        # x        : [B*N, h_dim]  original node embedding (passed via propagate kwarg)
        return self.drop(self.norm(F.relu(self.W(x + aggr_out))))   # residual + transform + norm + dropout


class SimpleGNN(nn.Module):
    # full model: encoder → message passing × 2 (per timestep) → temporal attention → global attention → decoder
    def __init__(self, in_dim=11, h_dim=96, edge_dim=1,
                 edge_h=16, n_mp=2, n_heads=4, w_out=24, dropout=0.2):
        super().__init__()
        self.h_dim   = h_dim
        self.n_wells = 4              # number of wells to predict
        self.w_out   = w_out

        self.encoder = nn.Sequential(                                                   # projects node features to hidden dim
            nn.Linear(in_dim, h_dim), nn.LayerNorm(h_dim), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(h_dim,  h_dim), nn.LayerNorm(h_dim), nn.ReLU(), nn.Dropout(dropout),
        )

        self.mp_layers = nn.ModuleList([                                                # stack of graph conv layers
            EdgeWeightedConv(h_dim, edge_dim, edge_h, dropout) for _ in range(n_mp)
        ])

        self.time_attn = nn.Linear(h_dim, 1)                                           # scores each timestep for temporal pooling

        self.global_attn = nn.MultiheadAttention(h_dim, n_heads, dropout=dropout,      # lets all nodes attend to each other
                                                  batch_first=True)
        self.norm1 = nn.LayerNorm(h_dim)    # layer norm after global attention
        self.ff    = nn.Sequential(         # feed-forward block inside transformer-style layer
            nn.Linear(h_dim, h_dim * 2), nn.ReLU(), nn.Dropout(dropout), nn.Linear(h_dim * 2, h_dim)
        )
        self.norm2 = nn.LayerNorm(h_dim)    # layer norm after feed-forward

        self.node_embed = nn.Embedding(4, h_dim)   # one learned identity vector per well (gives each well a unique bias)

        self.decoder = nn.Sequential(              # maps each well embedding to 24h forecast
            nn.Linear(h_dim, h_dim), nn.LayerNorm(h_dim), nn.ReLU(), nn.Dropout(dropout),
            nn.Linear(h_dim, w_out),
        )

    def forward(self, x, ea_seq, edge_index):
        # x        : [B, 72, N, 11]  node features over input window
        # ea_seq   : [B, 72, E, 1]   edge features over input window
        # edge_index : [2, E]        fixed graph topology (same for all batches)
        B, W, N, F = x.shape
        E = edge_index.shape[1]                                   # number of directed edges

        ei_batch = make_batch_edge_index(edge_index, B, N)        # [2, B*E] — one big graph per batch

        h_seq = []   # will collect node embeddings for each timestep
        for t in range(W):
            ht   = self.encoder(x[:, t].reshape(B * N, F))        # encode all nodes at time t → [B*N, h_dim]
            ea_t = ea_seq[:, t].reshape(B * E, 1)                 # edge features at time t → [B*E, 1]
            for mp in self.mp_layers:
                ht = mp(ht, ei_batch, ea_t)                        # apply graph convolution
            h_seq.append(ht.reshape(B, N, -1))                    # store result [B, N, h_dim]

        h_seq = torch.stack(h_seq, dim=1)                         # [B, 72, N, h_dim]

        scores = torch.softmax(self.time_attn(h_seq), dim=1)      # [B, 72, N, 1] attention weights over time
        h      = (h_seq * scores).sum(dim=1)                      # [B, N, h_dim] weighted average over timesteps

        attn_out, _ = self.global_attn(h, h, h)                   # all nodes attend to each other
        h = self.norm1(h + attn_out)                               # residual + norm
        h = self.norm2(h + self.ff(h))                             # feed-forward + residual + norm

        h_wells  = h[:, WELL_IDX, :]                              # select only the 4 well nodes [B, 4, h_dim]
        node_ids = torch.arange(4, device=x.device)               # indices 0,1,2,3 for the 4 wells
        h_wells  = h_wells + self.node_embed(node_ids)            # add learned identity per well

        out = self.decoder(h_wells)                                # [B, 4, 24]
        return out.permute(0, 2, 1)                                # [B, 24, 4] — match target shape


model = SimpleGNN(
    in_dim=IN_DIM, h_dim=H, edge_dim=EDGE_DIM,
    edge_h=16, n_mp=2, n_heads=4, w_out=W_OUT, dropout=DROPOUT,
).to(device)

print(f'Parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}')
```

```python
EPOCHS   = 300
LR       = 1e-3
PATIENCE = 40
SMOOTH_W = 5

optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=5e-4)   # adam with l2 regularisation
criterion = nn.MSELoss()                                                       # mean squared error loss
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(                        # halve lr if val loss stagnates
    optimizer, mode='min', factor=0.5, patience=20, verbose=True
)

best_val_smooth = float('inf')   # track best smoothed val loss for early stopping
patience_cnt    = 0              # counts epochs without improvement
train_losses, val_losses = [], []

for epoch in range(1, EPOCHS + 1):

    model.train()   # enable dropout
    total = 0.0
    for xb, eab, yb in train_loader:
        xb, eab, yb = xb.to(device), eab.to(device), yb.to(device)   # move batch to gpu if available
        optimizer.zero_grad()                                           # clear previous gradients
        loss = criterion(model(xb, eab, ei), yb)                       # forward pass + loss
        loss.backward()                                                 # backpropagation
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)        # gradient clipping to avoid explosion
        optimizer.step()                                                # update weights
        total += loss.item() * len(xb)
    train_loss = total / len(train_loader.dataset)   # average loss over all training samples

    model.eval()   # disable dropout
    with torch.no_grad():
        total = sum(
            criterion(model(xb.to(device), eab.to(device), ei), yb.to(device)).item() * len(xb)
            for xb, eab, yb in val_loader
        )
    val_loss = total / len(val_loader.dataset)

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    scheduler.step(val_loss)   # adjust lr based on val loss

    val_smooth = np.mean(val_losses[-SMOOTH_W:]) if len(val_losses) >= SMOOTH_W else val_loss   # smoothed val for early stop

    if epoch % 10 == 0:
        print(f'Epoch {epoch:3d} | train={train_loss:.4f}  val={val_loss:.4f}'
              f'  smooth={val_smooth:.4f}  lr={optimizer.param_groups[0]["lr"]:.2e}')

    if val_smooth < best_val_smooth:
        best_val_smooth = val_smooth   # new best → save checkpoint
        patience_cnt    = 0
        torch.save(model.state_dict(), 'best_model.pt')
    else:
        patience_cnt += 1
        if patience_cnt >= PATIENCE:
            print(f'Early stop at epoch {epoch}')
            break

plt.figure(figsize=(10, 3))
plt.plot(train_losses, label='Train')
plt.plot(val_losses,   label='Val')
plt.xlabel('Epoch'); plt.ylabel('MSE (normalised)'); plt.legend(); plt.tight_layout(); plt.show()
```

```python
model.load_state_dict(torch.load('best_model.pt'))   # reload best checkpoint
model.eval()

preds, trues = [], []
with torch.no_grad():
    for xb, eab, yb in test_loader:
        preds.append(model(xb.to(device), eab.to(device), ei).cpu().numpy())   # run inference
        trues.append(yb.numpy())

preds = np.concatenate(preds)   # [S_test, 24, 4]
trues = np.concatenate(trues)
S_test = len(preds)

preds_real = y_scaler.inverse_transform(preds.reshape(-1, 4)).reshape(S_test, 24, 4)   # denormalise predictions
trues_real = y_scaler.inverse_transform(trues.reshape(-1, 4)).reshape(S_test, 24, 4)   # denormalise ground truth

well_names = ['Node 1', 'Node 5', 'Node 17', 'Node 18']
print(f"{'Well':<10}  {'MAE':>8}  {'RMSE':>8}  {'NMAE':>8}")
print('-' * 44)
for i, name in enumerate(well_names):
    mae  = np.mean(np.abs(preds_real[:, :, i] - trues_real[:, :, i]))             # mean absolute error
    rmse = np.sqrt(np.mean((preds_real[:, :, i] - trues_real[:, :, i]) ** 2))     # root mean squared error
    nmae = mae / np.mean(np.abs(trues_real[:, :, i])) * 100                       # normalised mae (%)
    print(f'{name:<10}  {mae:>8.4f}  {rmse:>8.4f}  {nmae:>7.2f}%')

total_pred = preds_real.sum(axis=2)   # sum over 4 wells to get total injection [S, 24]
total_true = trues_real.sum(axis=2)
mae_tot  = np.mean(np.abs(total_pred - total_true))
rmse_tot = np.sqrt(np.mean((total_pred - total_true) ** 2))
nmae_tot = mae_tot / np.mean(np.abs(total_true)) * 100
print(f'\n{"Total":<10}  {mae_tot:>8.4f}  {rmse_tot:>8.4f}  {nmae_tot:>7.2f}%')

# ── plot 3 test days ───────────────────────────────────────────────────────────
N_DAYS = 3
total_true_cont = np.concatenate([total_true[d] for d in range(N_DAYS)])   # stitch days together
total_pred_cont = np.concatenate([total_pred[d] for d in range(N_DAYS)])
t = np.arange(N_DAYS * 24)

fig, axes = plt.subplots(2, 1, figsize=(14, 7))
axes[0].plot(t, total_true_cont, label='Real',      lw=2, color='steelblue')
axes[0].plot(t, total_pred_cont, label='Predicted', lw=2, color='tomato', ls='--')
axes[0].fill_between(t, total_true_cont, total_pred_cont, alpha=0.15, color='tomato')
for d in range(1, N_DAYS):
    axes[0].axvline(d * 24, color='grey', lw=0.8, ls=':')
axes[0].set_title(f'g total — {N_DAYS} consecutive test days')
axes[0].set_ylabel('g [MNm3]'); axes[0].legend(); axes[0].grid(alpha=0.3)
axes[0].text(0.99, 0.97, f'MAE={mae_tot:.4f}  RMSE={rmse_tot:.4f}  NMAE={nmae_tot:.2f}%',
             transform=axes[0].transAxes, ha='right', va='top', fontsize=9,
             bbox=dict(facecolor='white', alpha=0.7))

residuals = total_pred_cont - total_true_cont   # prediction error at each hour
axes[1].bar(t, residuals, color=['tomato' if r > 0 else 'steelblue' for r in residuals], alpha=0.7)
axes[1].axhline(0, color='black', lw=1)
for d in range(1, N_DAYS):
    axes[1].axvline(d * 24, color='grey', lw=0.8, ls=':')
axes[1].set_title('Residuals (predicted − real)')
axes[1].set_ylabel('Error [MNm3]'); axes[1].grid(alpha=0.3, axis='y')
plt.tight_layout(); plt.show()
```
